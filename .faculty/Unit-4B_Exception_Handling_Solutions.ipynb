**Task 1**

Write a python program to demonstrate try-except block for handling exceptions.


```python
# Example solution: try-except block
try:
    a = 10
    b = 0
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

**Task 2**

Write a python program to handle multiple exceptions using multiple except blocks.


```python
# Example solution: multiple except blocks
list_data = [10, 20, 30]
try:
    index = int(input("Enter index: "))
    value = list_data[index]
    print("Value:", value)
except ValueError:
    print("Please enter a valid integer index.")
except IndexError:
    print("Index is out of range.")
```

**Task 3**

Write a python program to demonstrate try-except-else-finally blocks.


```python
# Example solution: try-except-else-finally
try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    result = a / b
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except ValueError:
    print("Please enter valid numbers.")
else:
    print("Result:", result)
finally:
    print("Program finished.")
```

**Task 4**

Write a python program to handle ZeroDivisionError and ValueError.


```python
# Example solution: ZeroDivisionError and ValueError
try:
    num = float(input("Enter numerator: "))
    den = float(input("Enter denominator: "))
    print("Quotient:", num / den)
except ZeroDivisionError:
    print("Denominator cannot be zero.")
except ValueError:
    print("Please enter valid numbers.")
```

**Task 5**

Write a python program to demonstrate raising exceptions using raise.


```python
# Example solution: raise an exception
try:
    age = int(input("Enter age: "))
    if age < 0:
        raise ValueError("Age cannot be negative.")
    print("Valid age:", age)
except ValueError as err:
    print("Error:", err)
```

**Task 6**

Write a python program to create and raise a custom exception.


```python
# Example solution: custom exception
class InvalidSpeedError(Exception):
    pass


try:
    speed = int(input("Enter speed (km/h): "))
    if speed < 0 or speed > 120:
        raise InvalidSpeedError("Speed must be between 0 and 120.")
    print("Speed accepted:", speed)
except InvalidSpeedError as err:
    print("Custom error:", err)
except ValueError:
    print("Please enter a valid integer.")
```

**Task 7**

Write a python program to validate user input using exception handling (robust input program for student marks: custom exception, ValueError, Pass/Fail, finally).


```python
# Example solution: robust input validation for student marks
class InvalidMarksError(Exception):
    pass


try:
    marks = int(input("Enter student marks: "))
    if marks < 0 or marks > 100:
        raise InvalidMarksError("Marks must be between 0 and 100")
    if marks >= 40:
        print("Result: Pass")
    else:
        print("Result: Fail")
except ValueError:
    print("Error: Please enter numbers only.")
except InvalidMarksError as e:
    print("Invalid Input:", e)
finally:
    print("Marks processing completed.")
```
