**Task 1**

Write a python program to create, access and update a dictionary.

```python
# Example solution: create, access and update a dictionary
student = {
    "name": "Anu",
    "age": 20,
    "course": "Python",
}
print("Student:", student)
print("Name:", student["name"])

student["age"] = 21
student["city"] = "Guntur"
print("Updated:", student)
```

**Task 2**

Write a python program to demonstrate dictionary methods (`keys`, `values`, `items`, `get`, `update`, `pop`).

```python
# Example solution: dictionary methods
data = {"a": 1, "b": 2, "c": 3}
print("Keys:", data.keys())
print("Values:", data.values())
print("Items:", data.items())
print("get b:", data.get("b"))
print("get z:", data.get("z"))

data.update({"b": 20, "d": 4})
print("After update:", data)

removed = data.pop("a")
print("Removed:", removed)
print("After pop:", data)
```

**Task 3**

Write a python program to count the frequency of words in a sentence using a dictionary.

```python
# Example solution: word frequency dictionary
sentence = "to be or not to be"
words = sentence.split()
freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print("Frequency:", freq)
```

**Task 4**

Write a python program to create and access a nested dictionary.

```python
# Example solution: nested dictionary
students = {
    "S1": {"name": "Ram", "marks": 85},
    "S2": {"name": "Sita", "marks": 90},
}
print("All:", students)
print("S1 name:", students["S1"]["name"])
print("S2 marks:", students["S2"]["marks"])
```

**Task 5**

Write a python program to iterate through a dictionary using loops.

```python
# Example solution: iterate dictionary with loops
scores = {"maths": 80, "science": 75, "english": 90}

print("Keys:")
for key in scores:
    print(key)

print("Values:")
for value in scores.values():
    print(value)

print("Items:")
for key, value in scores.items():
    print(key, "→", value)
```

**Task 6**

Write a python program to demonstrate list comprehension (squares, filtering even numbers).

```python
# Example solution: list comprehension
numbers = range(1, 11)

squares = [x * x for x in numbers]
evens = [x for x in numbers if x % 2 == 0]

print("Squares:", squares)
print("Evens:", evens)
```

**Task 7**

Write a python program to demonstrate dictionary comprehension.

```python
# Example solution: dictionary comprehension
squares = {x: x * x for x in range(1, 6)}
print("Squares dict:", squares)

even_squares = {x: x * x for x in range(1, 11) if x % 2 == 0}
print("Even squares:", even_squares)
```

**Task 8**

Write a python program to merge two dictionaries.

```python
# Example solution: merge two dictionaries
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged = {**dict1, **dict2}
print("Merged:", merged)
```

**Task 9**

Write a python program to manage student records (name, marks, grade) using a dictionary.

```python
# Example solution: student records dictionary
name = "Krishna"
marks = 86

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
else:
    grade = "D"

student = {"name": name, "marks": marks, "grade": grade}
print("Record:", student)
for key, value in student.items():
    print(f"{key}: {value}")
```

**Task 10**

Write a python program to sort a list of dictionaries based on a key.

```python
# Example solution: sort list of dictionaries
students = [
    {"name": "Anu", "marks": 75},
    {"name": "Ravi", "marks": 90},
    {"name": "Sita", "marks": 82},
]

by_marks = sorted(students, key=lambda d: d["marks"])
print("By marks ascending:", by_marks)

by_marks_desc = sorted(students, key=lambda d: d["marks"], reverse=True)
print("By marks descending:", by_marks_desc)
```
