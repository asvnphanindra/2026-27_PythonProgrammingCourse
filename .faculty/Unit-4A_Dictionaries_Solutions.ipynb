Example solutions for Unit 4A dictionary tasks.

**Task 1**

Write a Python program that creates a student dictionary (name, age, course), updates age, and adds city.

```python
# Example solution: create, access and update a dictionary
dict_data = {
    "name": "Anu",
    "age": 20,
    "course": "Python",
}
print("Student:", dict_data)
print("Name:", dict_data["name"])

dict_data["age"] = 21
dict_data["city"] = "Guntur"
print("Updated:", dict_data)
```

**Task 2**

Write a Python program that uses keys, values, items, get, update and pop on a sample dictionary.

```python
# Example solution: dictionary methods
dict_data = {"a": 1, "b": 2, "c": 3}
print("Keys:", dict_data.keys())
print("Values:", dict_data.values())
print("Items:", dict_data.items())
print("get b:", dict_data.get("b"))
print("get z:", dict_data.get("z"))

dict_data.update({"b": 20, "d": 4})
print("After update:", dict_data)

removed = dict_data.pop("a")
print("Removed:", removed)
print("After pop:", dict_data)
```

**Task 3**

Write a Python program that counts how many times each word appears in a sentence.

```python
# Example solution: word frequency dictionary
sentence = "to be or not to be"
words = sentence.split()
dict_data = {}

for word in words:
    dict_data[word] = dict_data.get(word, 0) + 1

print("Frequency:", dict_data)
```

**Task 4**

Write a Python program that stores students in a nested dictionary and prints name and marks.

```python
# Example solution: nested dictionary
dict_data = {
    "S1": {"name": "Ram", "marks": 85},
    "S2": {"name": "Sita", "marks": 90},
}
print("All:", dict_data)
print("S1 name:", dict_data["S1"]["name"])
print("S2 marks:", dict_data["S2"]["marks"])
```

**Task 5**

Write a Python program that prints keys, values and items of a subject-marks dictionary using loops.

```python
# Example solution: iterate dictionary with loops
dict_data = {"maths": 80, "science": 75, "english": 90}

print("Keys:")
for key in dict_data:
    print(key)

print("Values:")
for value in dict_data.values():
    print(value)

print("Items:")
for key, value in dict_data.items():
    print(key, "→", value)
```

**Task 6**

Write a Python program that builds squares of 1–10 and even numbers using list comprehension.

```python
# Example solution: list comprehension
list_data = range(1, 11)

list_squares = [x * x for x in list_data]
list_evens = [x for x in list_data if x % 2 == 0]

print("Squares:", list_squares)
print("Evens:", list_evens)
```

**Task 7**

Write a Python program that builds a dictionary of numbers and their squares using dictionary comprehension.

```python
# Example solution: dictionary comprehension
dict_data = {x: x * x for x in range(1, 6)}
print("Squares dict:", dict_data)

dict_even_squares = {x: x * x for x in range(1, 11) if x % 2 == 0}
print("Even squares:", dict_even_squares)
```

**Task 8**

Write a Python program that merges two dictionaries with update() without changing the original.

```python
# Example solution: merge two dictionaries with update()
dict_data = {"a": 1, "b": 2}
dict_extra = {"b": 3, "c": 4}

dict_merged = dict_data.copy()
dict_merged.update(dict_extra)
print("Merged:", dict_merged)
```

**Task 9**

Write a Python program that stores a student's name, marks and grade in a dictionary and prints the record.

```python
# Example solution: student records dictionary
name = "Krishna"
marks = 86

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
else:
    grade = "D"

dict_data = {"name": name, "marks": marks, "grade": grade}
print("Record:", dict_data)
for key, value in dict_data.items():
    print(f"{key}: {value}")
```

**Task 10**

Write a Python program that sorts a list of student dictionaries by marks (ascending and descending).

```python
# Example solution: sort list of dictionaries
list_data = [
    {"name": "Anu", "marks": 75},
    {"name": "Ravi", "marks": 90},
    {"name": "Sita", "marks": 82},
]

list_by_marks = sorted(list_data, key=lambda d: d["marks"])
print("By marks ascending:", list_by_marks)

list_by_marks_desc = sorted(list_data, key=lambda d: d["marks"], reverse=True)
print("By marks descending:", list_by_marks_desc)
```
