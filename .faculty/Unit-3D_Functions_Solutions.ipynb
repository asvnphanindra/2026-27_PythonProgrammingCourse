**Task 1 — Rectangle area and perimeter**

Define a function that takes length and breadth, returns area and perimeter.


```python
def rectangle_area_perimeter(length, breadth):
    area = length * breadth
    perimeter = 2 * (length + breadth)
    return area, perimeter

length = 5
breadth = 3
area, perimeter = rectangle_area_perimeter(length, breadth)
print("Area:", area)
print("Perimeter:", perimeter)
```

**Task 2 — Celsius to Fahrenheit**

Define a function that converts Celsius to Fahrenheit and returns the result.


```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

celsius = 37
fahrenheit = celsius_to_fahrenheit(celsius)
print("Fahrenheit:", fahrenheit)
```

**Task 3 — Marks total and average**

Define a function that takes three subject marks and returns total and average.


```python
def total_and_average(mark1, mark2, mark3):
    total = mark1 + mark2 + mark3
    average = total / 3
    return total, average

m1, m2, m3 = 80, 70, 90
total, average = total_and_average(m1, m2, m3)
print("Total:", total)
print("Average:", average)
```

**Task 4 — Simple interest with default arguments**

Define a function for simple interest with parameters principal, time, and rate (default rate = 5). Call it once with the default rate and once with an explicit rate.


```python
def simple_interest(principal, time, rate=5):
    return (principal * time * rate) / 100

print("SI (default rate):", simple_interest(1000, 2))
print("SI (rate=8):", simple_interest(1000, 2, 8))
```

**Task 4a — Keyword vs positional arguments**

Define divide(a, b) that returns a / b. Call with keyword args divide(b=10, a=20) and positional args divide(10, 20). Print both results.


```python
def divide(a, b):
    return a / b

result = divide(b=10, a=20)  # keyword arguments
print(result)  # 2.0

result = divide(10, 20)  # positional arguments
print(result)  # 0.5
```

**Task 5 — Sum using *args**

Define a function using *args that returns the sum of any number of values.


```python
def sum_values(*args):
    total = 0
    for value in args:
        total += value
    return total

print(sum_values(1, 2, 3))
print(sum_values(10, 20, 30, 40))
```

**Task 6 — print_details with **kwargs**

Define print_details(**kwargs) that prints each key and value. Call as print_details(name="Krishna", age=26, city="Guntur").


```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Krishna", age=26, city="Guntur")
```

**Task 7 — shopping_cart with **products**

Define shopping_cart(**products) that prints each item with its price in ₹, then the total. Call as shopping_cart(apple=15, orange=12, mango=10).


```python
def shopping_cart(**products):
    total = 0
    print("Items Purchased:")
    for item, price in products.items():
        print(f"{item}: ₹{price}")
        total += price
    print(f"Total: ₹{total}")

shopping_cart(apple=15, orange=12, mango=10)
```

**Task 10 — Global counter increment**

Use a global counter and increment it inside a function with the global keyword.


```python
counter = 0

def increment():
    global counter
    counter = counter + 1
    print("Inside function, counter =", counter)

print("Before:", counter)
increment()
print("After:", counter)
```

**Task 11 — Bank balance: global vs local**

Use a global balance, a deposit function (with global), and a function with a local balance that does not update the account.


```python
balance = 1000

def deposit(amount):
    global balance
    balance = balance + amount
    print("After deposit, balance =", balance)

def show_local_balance():
    balance = 0  # local only
    print("Inside local function, balance =", balance)

print("Start:", balance)
deposit(500)
print("Account balance:", balance)
show_local_balance()
print("Account balance still:", balance)
```

**Task 12 — Factorial using recursion**

Find the factorial of a number using recursion.


```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

number = 5
print("Factorial:", factorial(number))
```

**Task 13 — Fibonacci series using recursion**

Generate the Fibonacci series using recursion.


```python
def fib(k):
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib(k - 1) + fib(k - 2)

n = 7
for i in range(n):
    print(fib(i), end=" ")
```

**Task 14 — GCD using recursion**

Find the GCD of two numbers using recursion.


```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

x, y = 48, 18
print("GCD:", gcd(x, y))
```

**Task 15 — Lambda with map, filter, reduce**

Demonstrate lambda functions with map(), filter(), and reduce().


```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))
total = reduce(lambda a, b: a + b, numbers)

print("Squares:", squares)
print("Evens:", evens)
print("Sum:", total)
```
