**Task 1**

Write a python program that defines a function to calculate the area and perimeter of a rectangle (length and breadth as parameters) and returns both values.


```python
def rectangle_area_perimeter(length, breadth):
    area = length * breadth
    perimeter = 2 * (length + breadth)
    return area, perimeter

length = 5
breadth = 3
area, perimeter = rectangle_area_perimeter(length, breadth)
print("Area:", area)
print("Perimeter:", perimeter)
```

**Task 2**

Write a python program that defines a function to convert temperature from Celsius to Fahrenheit (Celsius as parameter) and returns the result.

**Concept**

- Formula: Fahrenheit = (Celsius × 9/5) + 32
- Example: 37°C → (37 × 9/5) + 32 = 98.6°F


```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

celsius = 37
fahrenheit = celsius_to_fahrenheit(celsius)
print("Fahrenheit:", fahrenheit)
```

**Task 4**

Write a python program that defines a function to calculate simple interest with parameters principal, time, and rate (default rate = 5); call it using the default rate and also with an explicit rate.

**Concept — default arguments**

- A parameter can have a default value, e.g. `rate=5`
- If you omit it in the call, the default is used
- Formula: SI = (principal × time × rate) / 100


```python
def simple_interest(principal, time, rate=5):
    return (principal * time * rate) / 100

print("SI (default rate):", simple_interest(1000, 2))
print("SI (rate=8):", simple_interest(1000, 2, 8))
```

**Task 4a**

Write a python program that defines a function `divide(a, b)` returning `a / b`; call it with keyword arguments `divide(b=10, a=20)` and with positional arguments `divide(10, 20)`, and print both results.

**Concept — keyword vs positional arguments**

- Keyword: pass by name, e.g. `divide(b=10, a=20)` → order does not matter → `20 / 10 = 2.0`
- Positional: follow definition order, e.g. `divide(10, 20)` → `a=10`, `b=20` → `0.5`


```python
def divide(a, b):
    return a / b

result = divide(b=10, a=20)  # keyword arguments
print(result)  # 2.0

result = divide(10, 20)  # positional arguments
print(result)  # 0.5
```

**Task 5**

Write a python program that defines a function using `*args` to find the sum of any number of values.

**Concept — `*args`**

- `*args` collects any number of positional arguments into a tuple
- Example: `sum_values(1, 2, 3)` → `args` is `(1, 2, 3)`


```python
def sum_values(*args):
    total = 0
    for value in args:
        total += value
    return total

print(sum_values(1, 2, 3))
print(sum_values(10, 20, 30, 40))
```

**Task 6**

Write a python program that defines a function `print_details(**kwargs)` to print each key and value from `kwargs`; call it as `print_details(name="Krishna", age=26, city="Guntur")`.

**Concept — `**kwargs`**

- `**kwargs` collects keyword arguments into a dictionary
- Example: `name="Krishna", age=26` → `{"name": "Krishna", "age": 26}`


```python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Krishna", age=26, city="Guntur")
```

**Task 10**

Write a python program that uses a global counter and increments it inside a function using the `global` keyword.

**Concept — `global` keyword**

- To **change** a global variable inside a function, declare `global name` first
- Without `global`, assignment creates a local variable instead


```python
counter = 0

def increment():
    global counter
    counter = counter + 1
    print("Inside function, counter =", counter)

print("Before:", counter)
increment()
print("After:", counter)
```

**Task 11**

Write a python program for a simple bank balance: a global `balance`, a function to deposit (uses `global`), and a function that uses a local `balance` to show it does not update the account.

**Concept**

- `deposit` uses `global balance` so the account updates
- A function that only assigns `balance = ...` without `global` creates a local value; the account stays the same


```python
balance = 1000

def deposit(amount):
    global balance
    balance = balance + amount
    print("After deposit, balance =", balance)

def show_local_balance():
    balance = 0  # local only
    print("Inside local function, balance =", balance)

print("Start:", balance)
deposit(500)
print("Account balance:", balance)
show_local_balance()
print("Account balance still:", balance)
```

**Task 12**

Write a python program to find the factorial of a number using recursion.

**Concept — factorial**

- `n!` means multiply all whole numbers from `n` down to `1`
- Example: `5! = 5 × 4 × 3 × 2 × 1 = 120`
- Also: `0! = 1` and `1! = 1`

**Concept — recursion**

- A function that calls itself
- Needs a **base case** to stop
- Needs a **recursive case** that moves toward the base case
- Here: `n! = n × (n - 1)!`


```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

number = 5
print("Factorial:", factorial(number))
```

**Task 13**

Write a python program to generate Fibonacci series using recursion.

**Concept — Fibonacci series**

- Each term is the **sum of the previous two** terms
- Start with: `0, 1`
- Next terms: `0+1=1`, `1+1=2`, `1+2=3`, `2+3=5`, …
- First 7 terms: `0 1 1 2 3 5 8`

**Recursive rule**

- `fib(0) = 0`
- `fib(1) = 1`
- `fib(k) = fib(k-1) + fib(k-2)`


```python
def fib(k):
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib(k - 1) + fib(k - 2)

n = 7
for i in range(n):
    print(fib(i), end=" ")
```

**Task 14**

Write a python program to find the GCD of two numbers using recursion.

**Concept — GCD**

- GCD = Greatest Common Divisor = largest number that divides both numbers exactly
- Example: GCD of 48 and 18 is **6** (because 48=6×8 and 18=6×3)

**Euclidean algorithm with remainder**

1. Divide `a` by `b` and take remainder: `r = a % b`
2. Replace: `a ← b`, `b ← r`
3. Repeat until `b` becomes `0`
4. Then GCD is `a`

Example: `gcd(48, 18)`

| Step | a | b | a % b |
|------|---|---|-------|
| 1 | 48 | 18 | 12 |
| 2 | 18 | 12 | 6 |
| 3 | 12 | 6 | 0 |
| 4 | 6 | 0 | stop → GCD = 6 |

Recursive form: `gcd(a, b) = a` if `b == 0`, else `gcd(b, a % b)`


```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

x, y = 48, 18
print("GCD:", gcd(x, y))
```

**Task 15**

Write a python program to demonstrate lambda functions with map(), filter() and reduce().

**Concept — lambda, map, filter, reduce**

- `lambda`: short one-line function, e.g. `lambda x: x * x`
- `map`: apply a function to every item
- `filter`: keep items that match a condition
- `reduce`: combine items into one value (import from `functools`)


```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x * x, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))
total = reduce(lambda a, b: a + b, numbers)

print("Squares:", squares)
print("Evens:", evens)
print("Sum:", total)
```
