**Task 1**

Write a python program to count the number of vowels in a given string.

```python
# Example solution: count vowels in a string
text = "NRI Institute of Technology"
vowels = "aeiouAEIOU"
count = 0

for ch in text:
    if ch in vowels:
        count = count + 1

print("Number of vowels:", count)
```

**Task 2**

Write a python program to count the frequency of each character in a string (ignore spaces) and store it in a dictionary.

```python
# Example solution: character frequency dictionary
text = "Welcome to Python Programming"
freq = {}

for ch in text:
    if ch == " ":
        continue
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1

print(freq)
```

**Task 3**

Write a python program to calculate the sum of all elements in a list using iteration (do not use `sum()`).

```python
# Example solution: sum of list elements using an accumulator
numbers = [12, 25, 8, 15, 30, 10]
total = 0

for num in numbers:
    total = total + num

print("Total sum:", total)
```

**Task 4**

Write a python program to separate even and odd numbers from a list into two new lists.

```python
# Example solution: separate even and odd numbers into two lists
numbers = [12, 7, 25, 8, 16, 33, 40, 15]
even_numbers = []
odd_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```

**Task 5**

Write a python program to count how many positive numbers are present in a tuple.

```python
# Example solution: count positive numbers in a tuple
numbers = (10, -5, 0, 25, -12, 8, -3, 15)
count = 0

for num in numbers:
    if num > 0:
        count = count + 1

print("Positive numbers:", count)
```

**Task 6**

Write a python program to calculate the average of marks in a tuple and count how many subjects scored above the average.

```python
# Example solution: average marks and count of subjects above average
marks = (85, 72, 90, 65, 78, 95)
total = 0
count = 0

for m in marks:
    total = total + m
    count = count + 1

average = total / count

above = 0
for m in marks:
    if m > average:
        above = above + 1

print("Average marks:", round(average, 2))
print("Subjects above average:", above)
```

**Task 7**

Write a python program to count how many numbers in a set are greater than 10.

```python
# Example solution: count set elements greater than 10
numbers = {5, 12, 18, 7, 25, 9, 30}
count = 0

for num in numbers:
    if num > 10:
        count = count + 1

print("Numbers greater than 10:", count)
```

**Task 8**

Write a python program to create a new set containing the squares of all elements from a given set.

```python
# Example solution: create a set of squares
numbers = {2, 4, 6, 8, 10}
squares = set()

for num in numbers:
    squares.add(num * num)

print(squares)
```

**Task 9**

Write a python program to calculate the total marks of all students from a dictionary (do not use `sum()`).

```python
# Example solution: total marks from a dictionary
student_marks = {
    "Anil": 78,
    "Bhavana": 92,
    "Charan": 65,
    "Divya": 85
}
total = 0

for marks in student_marks.values():
    total = total + marks

print("Total marks:", total)
```

**Task 10**

Write a python program to find the class average, print students who scored above average, and count students who scored below 50.

```python
# Example solution: class average, students above average, count below 50
student_marks = {
    "Anil": 78,
    "Bhavana": 92,
    "Charan": 45,
    "Divya": 88,
    "Kiran": 36,
    "Meena": 95
}

total = 0
count = 0

for marks in student_marks.values():
    total = total + marks
    count = count + 1

average = total / count
print("Class average:", round(average, 2))

print("\nStudents above average:")
below_50 = 0

for name, marks in student_marks.items():
    if marks > average:
        print(name)
    if marks < 50:
        below_50 = below_50 + 1

print("\nStudents below 50:", below_50)
```
