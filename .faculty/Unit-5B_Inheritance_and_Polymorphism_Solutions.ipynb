**Task 1**

Demonstrate single inheritance.

**File:** `Unit-5B_Task-1_Single_Inheritance.py`


```python
# Example solution: single inheritance
class Person:
    def __init__(self, name):
        self.name = name

    def show(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def show_course(self):
        print("Course:", self.course)


s = Student("Anu", "Python")
s.show()
s.show_course()
```

**Task 2**

Demonstrate multiple and multilevel inheritance.

**File:** `Unit-5B_Task-2_Multiple_Multilevel_Inheritance.py`


```python
# Example solution: multilevel and multiple inheritance
# Multilevel: A -> B -> C
class A:
    def feature_a(self):
        print("Feature A")


class B(A):
    def feature_b(self):
        print("Feature B")


class C(B):
    def feature_c(self):
        print("Feature C")


# Multiple: Child inherits Father and Mother
class Father:
    def skill_f(self):
        print("Father skill")


class Mother:
    def skill_m(self):
        print("Mother skill")


class Child(Father, Mother):
    pass


obj = C()
obj.feature_a()
obj.feature_b()
obj.feature_c()

ch = Child()
ch.skill_f()
ch.skill_m()
```

**Task 3**

Demonstrate method overriding using inheritance.

**File:** `Unit-5B_Task-3_Method_Overriding.py`


```python
# Example solution: method overriding
class Animal:
    def sound(self):
        print("Some sound")


class Dog(Animal):
    def sound(self):
        print("Bark")


a = Animal()
d = Dog()
a.sound()
d.sound()
```

**Task 4**

Demonstrate polymorphism using method overriding.

**File:** `Unit-5B_Task-4_Polymorphism_Method_Overriding.py`


```python
# Example solution: polymorphism with method overriding
class Shape:
    def area(self):
        print("Undefined")


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        print("Square:", self.side * self.side)


class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        print("Circle:", 3.14 * self.r * self.r)


shapes = [Square(4), Circle(3)]
for s in shapes:
    s.area()
```

**Task 5**

Demonstrate operator overloading.

**File:** `Unit-5B_Task-5_Operator_Overloading.py`


```python
# Example solution: operator overloading
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)
```
