# Unit 4C — File Handling (Solutions)

Example solutions for Unit 4C file handling tasks.


**Task 1**

Write a Python program that reads sample.txt and displays its content.


```python
# Example solution: read a text file
filename = "sample.txt"
with open(filename, "r") as file:
    content = file.read()
print(content)
```

**Task 2**

Write a Python program that writes a line of text to output.txt.


```python
# Example solution: write to a text file
filename = "output.txt"
text = "Learning file handling in Python."
with open(filename, "w") as file:
    file.write(text)
print("Data written to", filename)
```

**Task 3**

Write a Python program that counts lines, words and characters in a text file.


```python
# Example solution: count lines, words, characters
filename = "sample.txt"
with open(filename, "r") as file:
    text = file.read()

line_count = len(text.splitlines()) if text else 0
word_count = len(text.split())
char_count = len(text)

print("Lines:", line_count)
print("Words:", word_count)
print("Characters:", char_count)
```

**Task 4**

Write a Python program that copies the content of sample.txt into copy.txt.


```python
# Example solution: copy file content
source = "sample.txt"
destination = "copy.txt"
with open(source, "r") as src:
    content = src.read()
with open(destination, "w") as dest:
    dest.write(content)
print("Copied from", source, "to", destination)
```

**Task 5**

Write a Python program that replaces a word in a text file (for example Python with Java).


```python
# Example solution: search and replace in a file
filename = "sample.txt"
old_word = "Python"
new_word = "Java"

with open(filename, "r") as file:
    content = file.read()

updated = content.replace(old_word, new_word)

with open(filename, "w") as file:
    file.write(updated)

print("Replaced", old_word, "with", new_word)
```
