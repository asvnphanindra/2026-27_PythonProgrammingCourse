**Task 1**

Create a class and object, and access attributes/methods.

**File:** `Unit-5A_Task-1_Class_and_Object.py`


```python
# Example solution: class and object
class Student:
    def greet(self):
        print("Hello from", self.name)


student = Student()
student.name = "Anu"
student.age = 20
print(student.name, student.age)
student.greet()
```

**Task 2**

Demonstrate constructor (`__init__`) in a class.

**File:** `Unit-5A_Task-2_Constructor_Init.py`


```python
# Example solution: constructor __init__
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name, self.age)


s1 = Student("Ravi", 21)
s1.display()
```

**Task 3**

Demonstrate instance methods, class methods and static methods.

**File:** `Unit-5A_Task-3_Instance_Class_Static_Methods.py`


```python
# Example solution: instance, class, and static methods
class Demo:
    college = "NRIU"

    def set_name(self, name):
        self.name = name
        print("Instance:", self.name)

    @classmethod
    def show_college(cls):
        print("Class:", cls.college)

    @staticmethod
    def info():
        print("Static: utility message")


obj = Demo()
obj.set_name("Anu")
Demo.show_college()
Demo.info()
```

**Task 4**

Demonstrate encapsulation using private and protected attributes.

**File:** `Unit-5A_Task-4_Encapsulation_Private_Protected.py`


```python
# Example solution: private and protected attributes
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._branch = "Main"  # protected (convention)
        self.__balance = balance  # private (name-mangled)

    def get_balance(self):
        return self.__balance


acc = BankAccount("Sita", 1000)
print(acc.owner, acc._branch)
print("Balance:", acc.get_balance())
```

**Task 5**

Demonstrate getter and setter methods.

**File:** `Unit-5A_Task-5_Getter_Setter_Methods.py`


```python
# Example solution: getter and setter methods
class Student:
    def __init__(self):
        self.__marks = 0

    def get_marks(self):
        return self.__marks

    def set_marks(self, value):
        if 0 <= value <= 100:
            self.__marks = value
        else:
            print("Invalid marks")


s = Student()
s.set_marks(85)
print("Marks:", s.get_marks())
s.set_marks(150)
```
